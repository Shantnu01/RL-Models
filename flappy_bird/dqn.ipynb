```python
import torch;

import torch.nn as nn   
```

```python
class DQN(nn.Module):
    def __init__(self,state_size=12,action_size=2,hidden_dim=256):

        super(DQN, self).__init__()
        
        self.s_size=state_size;
        self.a_size=action_size;
        self.hidden_layer=hidden_dim;
        self.model=nn.Sequential(
           nn.Linear(self.s_size,2*self.hidden_layer),
           nn.ReLU(),
           nn.Linear(2*self.hidden_layer,self.hidden_layer),
           nn.ReLU(),
           
           nn.Linear(self.hidden_layer,self.a_size)
        )
    def forward(self,x):
        return self.model(x);
```
